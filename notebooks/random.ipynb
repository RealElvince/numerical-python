```python
import numpy as np
import csv

# Set random seed for reproducibility
np.random.seed(123)

# Step 1: Generate 1000 normally distributed scores with mean=70, std=10
scores = np.random.normal(loc=70, scale=10, size=1000)

# Step 2: Print statistics
print(f"Mean: {np.mean(scores):.1f}")
print(f"Standard Deviation: {np.std(scores):.2f}")
print(f"Min: {np.min(scores):.2f}")
print(f"Max: {np.max(scores):.2f}")

# Step 3: Adjust mean to 65, regenerate scores
adjusted_scores = np.random.normal(loc=65, scale=10, size=1000)
print(f"Adjusted Mean: {np.mean(adjusted_scores):.2f}")

# Step 4: Save adjusted scores to CSV
with open('simulated_scores.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Scores'])
    for score in adjusted_scores:
        writer.writerow([score])

# Step 5: Confirmation message
print("Saved to simulated_scores.csv")
```
